## Problem 1: Arithmetic Operations ( Using Functions )

Write a Python program that performs various arithmetic operations on two numbers entered by the user.

1. Take **two numbers as input from the user**.

2. Perform the following **arithmetic operations**:
   - Addition
   - Subtraction
   - Multiplication
   - Division

3. The program must **handle division by zero gracefully**.If the user attempts to divide by zero, the program should display an appropriate message instead of producing an error.

4. If you are familiar with the concept of **function definition in Python**, implement each operation using **separate functions**.

5. Each function should:
   - Take the necessary inputs as parameters
   - Perform the required operation
   - Return or display the result.



````python
def add(x, y):
    return x + y

def sub(x, y):
    return x - y

def mul(x, y):
    return x * y

def div(x, y):
    return x / y

def main():
    x = int(input("x: "))
    y = int(input("y: "))
    op = input("Operator (+,-,*,/): ")

    match op:
        case '+': 
            print(f"\nAddition: {add(x, y)}")
        case '-':
            print(f"\nSubtraction: {sub(x, y)}")
        case '*':
            print(f"\nMulitplication: {mul(x, y)}")
        case '/':
            try:
                print(f"\nDivision: {div(x, y)}")
            except:
                print("\nError: Division by zero not possible!")
        case _:
            print("Invalid Operator!")


if __name__ == "__main__":
    main()
    
````

## Problem: Password Strength Checker

Write a Python program to check whether a given password is **strong or not** based on the following conditions.

### Password Requirements

A valid password should satisfy the following criteria:

1. The password must contain **at least 8 characters**.
2. The password must contain **at least one uppercase letter** (`A–Z`).
3. The password must contain **at least one lowercase letter** (`a–z`).
4. The password must contain **at least one digit** (`0–9`).
5. The password must contain **at least one special character** from the following set:

\[
! \ @ \ \# \ \$ \ \% \ \^ \ \& \ *
\]

### Task

Write a Python program that:

1. Takes a **password as input from the user**.
2. Checks whether the password satisfies the above conditions.
3. Based on the number of conditions satisfied, classify the password as:

| Conditions Satisfied | Password Strength |
|---------------------|------------------|
| 1–2 conditions | Weak |
| 3–4 conditions | Moderate |
| All 5 conditions | Strong |

### Example

**Input**

```
Enter Password: P@ss1234
```

**Output**

```
Password Strength: Strong
```

````python
import re

def check_password_strength(password):
    c = 0
    if len(password) >= 8:
        c += 1

    if re.search(r"[A-Z]+", password):
        c += 1

    if re.search(r"[a-z]+", password):
        c += 1

    if re.search(r"\W+", password):
        c += 1

    if re.search(r"\d+", password):
        c += 1

    if c >= 1 and c <= 2:
        return "Weak"
    elif c >= 3 and c <= 4:
        return "Moderate"
    else:
        return "Strong"
    
def main():
    password = input("Password: ")
    # password = "sdfs&dDF12"

    print(check_password_strength(password))
    

if __name__ == "__main__":
    main()
````

## Problem 7: Triangle Validation and Classification

A triangle is formed by three sides. However, not every set of three numbers can form a valid triangle.  
The **Triangle Inequality Theorem** states that the sum of the lengths of any two sides of a triangle must be **greater than the third side**.

Mathematically,

\[
a + b > c
\]

\[
a + c > b
\]

\[
b + c > a
\]

where \(a\), \(b\), and \(c\) represent the three sides of the triangle.

---

### Task

Write a Python program that performs the following steps:

1. Take the **three side lengths** of a triangle as input from the user.
2. Check whether the given sides satisfy the **triangle inequality condition**.
3. If the triangle is **not valid**, display an appropriate message.
4. If the triangle is **valid**, determine the type of triangle.

---

### Triangle Classification

If the triangle is valid, classify it based on the side lengths:

| Triangle Type | Condition |
|---------------|-----------|
| Equilateral Triangle | All three sides are equal |
| Isosceles Triangle | Any two sides are equal |
| Scalene Triangle | All three sides are different |

---

### Example 1

**Input**

```
Enter side a: 5
Enter side b: 5
Enter side c: 5
```

**Output**

```
Valid Triangle
Type: Equilateral Triangle
```

---

### Example 2

**Input**

```
Enter side a: 4
Enter side b: 4
Enter side c: 6
```

**Output**

```
Valid Triangle
Type: Isosceles Triangle
```

---

### Example 3

**Input**

```
Enter side a: 2
Enter side b: 3
Enter side c: 10
```

**Output**

```
Not a valid triangle
```



````python
def is_valid_triangle(a, b, c):
    if a + b > c and a + c > b and b + c > a:
        return True
    else:
        return False

def is_isosceles_triangle(a, b, c):
    if a == b or a == c or b == c:
        return True
    else:
        return False
    
def is_equilateral_triangle(a, b, c):
    if a == b and a == c and b == c:
        return True
    else:
        return False
    
def is_scalene_triangle(a, b, c):
    if a != b and a != c and b != c:
        return True
    else:
        return False
    
def main():
    a = float(input("Side a: "))
    b = float(input("Side b: "))
    c = float(input("Side c: "))

    if is_valid_triangle(a, b, c):
        print("Valid Triangle!")
        if is_isosceles_triangle(a, b, c):
            print("Isosceles Triangle!")
        elif is_equilateral_triangle(a, b, c):
            print("Equilateral Triangle!")
        elif is_scalene_triangle(a, b, c):
            print("Scalene Triangle!")
    else:
        print("Invalid Triangle!")

if __name__ == "__main__":
    main()
````
