Q4. Write a Python program to iterate through a string and count the number of vowels present in it.

```python
txt = input("Text: ").lower()
c = 0

for i in txt:
    if i in ['a', 'e', 'i', 'o', 'u']:
        c += 1

print(f"Number of Vowels: {c}")
```

Q11. Write a Python program to create a set and iterate through it to print all unique elements.

```python
x = set()
n = int(input("n: "))

for _ in range(n):
    x.add(int(input()))

for i in x:
    print(i)
```

Q20. Write a Python program to implement a generator function to generate Fibonacci numbers up to N.

```python
def fibonacci(n):
    a, b = 0, 1
    for i in range(n):
        print(a)
        c = a + b
        a = b
        b = c

fibonacci(10)
```

Q21. Write a Python program to find two numbers in a list that add up to a given target value (Two Sum problem).

```python
def two_sum(lst, target):
    for i in range(len(lst)):
        for j in range(len(lst)):
            if i == j:
                continue
            else:
                if lst[i] + lst[j] == target:
                    return [lst[i], lst[j]]


def main():
    lst = [10, 20, 29, 37, 25]
    print(two_sum(lst, 49))

if __name__ == "__main__":
    main()
```
