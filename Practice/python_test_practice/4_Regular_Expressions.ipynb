### Q1
**Problem:** Find all words in a string that end with the suffix 'ing'.

```python
import re 
text = "dancing prancing running 213.123.123.23"
output = re.findall(r"[a-zA-z]+ing", text)

print(output)
```

### Q2
**Problem:** Extract all valid IPv4 addresses from a log file text.

```python
output = re.findall(r"[0-2][0-5][0-5]\.[0-2][0-5][0-5]\.[0-2][0-5][0-5]\.[0-2][0-5][0-5]", text)
print(output)
```
