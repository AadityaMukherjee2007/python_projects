### Q1
**Problem:** Write a function `calculate_factorial(n)` that returns the factorial of a number.

```python
def calculate_factorial(n):
    if n == 0:
        return 1
    return n * calculate_factorial(n - 1)

def main():
    print(calculate_factorial(5))

if __name__ == "__main__":
    main()
```

### Q2
**Problem:** Write a function `is_prime(n)` that checks if a number is prime.

```python
def is_prime(n):
    if n < 2:
        return False

    for i in range(2, int(n**0.5+1)):
        if n % i == 0:
            return False
    return True

def main():
    print(is_prime(1))
    print(is_prime(2))
    print(is_prime(4))
    print(is_prime(13))
    print(is_prime(49))

if __name__ == "__main__":
    main()
```

### Q3
**Problem:** Write a function `fibonacci_sequence(n)` that returns a list of the first n Fibonacci numbers.

```python
def fibonacci_sequence(n):
    a, b = 0, 1
    l = []
    for i in range(n):
        l.append(a)
        c = a + b
        a = b
        b = c
    return l

def main():
    print(fibonacci_sequence(10))

if __name__ == "__main__":
    main()
```

### Q4
**Problem:** Write a function `euclidean_distance(p1, p2)` that takes two tuples representing 2D points and returns the distance between them.

```python
def euclidean_distance(p1, p2):
    return ((p1[0] - p2[0])**2 + (p1[1] - p2[1])**2)**0.5

def main():
    print(euclidean_distance((1, 2), (2, 3)))

if __name__ == "__main__":
    main()
```

### Q5
**Problem:** Write a function `is_palindrome(s)` that ignores spaces and case to check if a string is a palindrome.

```python
def is_palindrome(s):
    s = s.lower().replace(" ", "")
    return s[::-1] == s

def main():
    print(is_palindrome("mom"))
    print(is_palindrome("father"))
    print(is_palindrome("Madam"))

if __name__ == "__main__":
    main()
```

### Q6
**Problem:** Write a function `matrix_transpose(matrix)` that takes a 2D list (matrix) and returns its transpose.

```python
def matrix_transpose(matrix):
    transpose = []
    for i in range(0, len(matrix[0])):
        row = []
        for j in range(0, len(matrix)):
            row.append(matrix[j][i])
        transpose.append(row)
    return transpose

def main():
    matrix = [
        [1, 2],
        [4, 5],
        [7, 8]
    ]

    matrix_t = matrix_transpose(matrix)
    print(matrix_t)

if __name__ == "__main__":
    main()
```

### Q7
**Problem:** Write a function `element_frequency(lst)` that returns a dictionary containing the frequency of each element in a list.

```python
def element_frequency(lst):
    freq = {}
    for i in lst:
        freq[i] = freq.get(i, 0) + 1
    return freq

def main():
    print(element_frequency([1, 2, 3, 1, 2, 5, 6, 4, 5, 6, 4, 1]))

if __name__ == "__main__":
    main()
```

### Q8
**Problem:** Write a function `compute_gcd(a, b)` to find the greatest common divisor of two numbers.

```python
def compute_gcd(a, b):
    while a != 0:
        r = b % a
        b = a
        a = r
    return b

def main():
    print(compute_gcd(8, 28))

if __name__ == "__main__":
    main()
        
```

### Q9
**Problem:** Write a function `standard_deviation(data)` that calculates the standard deviation of a list of numbers.

```python
def standard_deviation(data):
    mean = sum(data) / len(data)
    deviation = 0
    for i in data:
        deviation += (i - mean) ** 2

    return (deviation / len(data)) ** 0.5

def main():
    print(f"Deviation: {standard_deviation([1, 2, 3, 4, 2, 1, 2, 3, 5, 6, 4, 3, 4, 2, 1, 5])}")

if __name__ == "__main__":
    main()
```

### Q10
**Problem:** Write a function `unique_elements(lst)` that takes a list and returns a new list with unique elements, preserving the original order.

```python
def unique_elements(lst):
    unique = set()
    for i in lst:
        if i not in unique:
            unique.add(i)
    return list(unique)

def main():
    print(unique_elements([1, 2, 3, 2, 3, 5, 4, 3, 2, 1, 5]))

if __name__ == "__main__":
    main()
```
