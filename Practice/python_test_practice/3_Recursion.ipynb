### Q1
**Problem:** Write a recursive function to find the sum of the first n natural numbers.

```python
def sum(n):
    if n == 0:
        return 0
    return n + sum(n - 1)

sum(100)
```

### Q2
**Problem:** Write a recursive function `power(x, n)` to calculate x raised to the power n.

```python
def pow(x, n):
    if n == 0:
        return 1
    return x * pow(x, n - 1)

pow(3, 4)
```

### Q3
**Problem:** Write a recursive function to reverse a string.

```python
def reverse(str, i):
    if i == -1:
        return ""
    return str[i] + reverse(str, i - 1)

str = "Hello"
reverse(str, len(str) - 1)
```

### Q4
**Problem:** Write a recursive function to count the number of digits in a positive integer.

```python
def count(n):
    if n == 0:
        return 0
    return 1 + count(n // 10)

count(12313)
```

### Q5
**Problem:** Write a recursive function to find the sum of the digits of a number.

```python
def sum_of_digits(n):
    if n == 0:
        return 0

    return n % 10 + sum_of_digits(n // 10)

sum_of_digits(1234)
```

### Q10
**Problem:** Implement binary search recursively on a sorted list.

```python
lst = [1, 2, 3, 4, 5, 6, 7]

def search(lst, target, l=0, u=None):
    if u == None:
        u = len(lst) -1
    elif u < l:
        return -1
    
    mid = (u + l) // 2

    if lst[mid] < target:
        return search(lst, target, mid + 1, u)
    elif lst[mid] > target:
        return search(lst, target, l, mid - 1)
    else:
        return mid

print(search(lst, 5)) 
```
