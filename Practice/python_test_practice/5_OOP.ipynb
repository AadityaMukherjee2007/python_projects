### Q1
**Problem:** Create a `Student` class with attributes `name` and `roll_no`, and a method `display_details()`.

```python
class Student:
    def __init__(self, name, roll_no):
        self.name = name
        self.roll_no = roll_no

    def display(self):
        print(f"Name: {self.name}\nRoll: {self.roll_no}")

std1 = Student("John", 1)
std2 = Student("Luke", 2)
std3 = Student("Peter", 9)

std1.display()
std2.display()
std3.display()
```

### Q2
**Problem:** Create a `Rectangle` class initialized with length and width, featuring methods to calculate `area()` and `perimeter()`.

```python
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * (self.length + self.width)


rect = Rectangle(2, 3)
print(f"Area: {rect.area()}")
print(f"Perimeter: {rect.perimeter()}")
```

### Q3
**Problem:** Create a `BankAccount` class with an initial balance, and methods `deposit()` and `withdraw()` that update and print the balance.

```python
class BankAccount:
    def __init__(self, balance):
        self.balance = balance

    def display_balance(self):
        print(f"Balance: {self.balance}")

    def deposit(self, amount):
        self.balance += amount
        self.display_balance()

    def withdraw(self, amount):
        if amount > self.balance:
            raise Exception("Invalid Transaction")
        else:
            self.balance -= amount
            self.display_balance()

    

b = BankAccount(10000)
b.deposit(1000)
b.withdraw(11000)
```

### Q5
**Problem:** Implement a `Vector` class representing a 2D mathematical vector. Overload the `+` operator (`__add__`) to allow adding two vectors together.

```python
class Vector: 
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)
    
    def __str__(self):
        return f"({self.x}, {self.y})"
    

v1 = Vector(1, 2)            
v2 = Vector(2, 3)
v3 = v1 + v2

print(v1)
print(v2)
print(v3)
```
