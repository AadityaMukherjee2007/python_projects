### 1. Lists and Iteration
**Q. Write a program to check whether a given list contains any repeating elements.**
- Example input: `[5-8]` -> Output: `False`
- Example input: `[5-7]` -> Output: `True`

```python
# Write your code here
def is_repeating(lst):
    for i in range(len(lst)):
        for j in range(i + 1, len(lst)):
            if lst[i] == lst[j]:
                return True
    return False

is_repeating([1, 2, 1, 4])
#is_repeating([1, 2, 3, 4])
```

### 2. Strings
**Q. A string is said to be a rotation of another string if it can be obtained by shifting characters from the beginning of the string to the end without changing their order.**
Given two strings s1 and s2 determine whether s2 is a rotation of s1.
- Example input: `s1 = "abcd"`, `s2 = "cdab"` -> Output: `True`

```python
# Write your code here
def is_rotation(s1, s2):
    return s1[len(s1)//2:] + s1[:len(s1)//2] == s2

is_rotation("abcd", "cdab")
```

### 3. Regular Expressions
**Q. Write a program that takes a user id from the user and checks whether the id follows the given pattern:**
- the id starts with the string user followed by 3 digits
- followed by one special character followed by any number of characters

```python
import re
# Write your code here
text = "user123@f"

if re.search(r"^user\d{3}\W\w+", text):
    print("Valid")
else:
    print("Invalid")
```

### 4. Map, Filter, and Lambda
**Q. You are given a list of student records where `student = [("701", "std1", 50), ...]`**
Write a user defined function named `is_eligible` that returns true if the student's mark is less than 30.
Use the above function with `filter` operation to extract the ids of the eligible students.
Using `lambda` function with `map` create a new list with student's id and update its marks where students get a grace marks of 5.

```python
students = [
    ("701", "std1", 57), 
    ("702", "std2", 51), 
    ("703", "std3", 28), 
    ("706", "std6", 12)
]

# Write your code here
def is_eligible(rec):
    return rec[2] < 30

output = list(map(lambda x: x[2] + 5, filter(is_eligible, students)))
print(output)
```

### 5. Recursion
**Q. Summation of n natural numbers using recursion.**

```python
# Write your code here
def sum_nat(n):
    if n == 0:
        return 0

    return n + sum_nat(n - 1)

sum_nat(100)
```
