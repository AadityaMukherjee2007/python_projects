# Regex Practice Notebook

```python
import re
text = '''User1 192.168.1.1 email test@example.com phone 9876543210 date 12-05-2023'''
print(text)
```

Extract emails, IPs, dates, phone numbers

```python
# solve here
emails = re.findall(r"\w+@\w+\.\w+", text)
dates = re.findall(r"[0-1]?[1-9]-[0-3][0-9]-[0-9]{4}", text)
phone = re.findall(r"[0-9]{10}", text)
ip = re.fullmatch(r"\b((25[0-5]|2[0-4]\d|1\d\d|[1-9]?\d)\.){3}(25[0-5]|2[0-4]\d|1\d\d|[1-9]?\d)\b", text)

print(f"Emails: {emails}")
print(f"Dates: {dates}")
print(f"Phone Number: {phone}")
print(f"IP: {ip}")
```
