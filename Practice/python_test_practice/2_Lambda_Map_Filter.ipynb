### Q1
**Problem:** Create a lambda function that takes one argument and adds 15 to it.

```python
add = lambda x: x + 15
print(add(3))
```

### Q2
**Problem:** Create a lambda function that multiplies two arguments and returns the result.

```python
mul = lambda x, y: x * y
print(mul(3, 4))
```

### Q3
**Problem:** Use `map()` and a lambda function to square all numbers in a given list.

```python
lst = [i for i in range(1, 10)]
print(lst)

sq_lst = map(lambda x: x**2, lst)
print(list(sq_lst))
```

### Q4
**Problem:** Use `filter()` to extract only the even numbers from a list of integers.

```python
even_lst = filter(lambda x: x % 2 == 0, lst)
print(list(even_lst))
```

### Q5
**Problem:** Use `filter()` to extract all positive numbers from a list containing both positive and negative integers.

```python
lst = [1, -2, 4, -6, 4, 2, -2, 3, 5, -9]
postive_list = list(filter(lambda x: x >= 0, lst))

print(postive_list)
```

### Q6
**Problem:** Use `map()` to convert a list of strings to uppercase.

```python
lst = ["hello", "World", "this is programming", "WIth", "Python"]
output = list(map(lambda x: x.upper(), lst))
print(output)
```

### Q7
**Problem:** Use `filter()` and a lambda function to find all palindromes in a list of words.

```python
lst = ["mom", "Madam", "Father", "God"]
output = filter(lambda x: x.lower() == x.lower()[::-1], lst)
print(list(output))
```

### Q8
**Problem:** Use `map()` to extract the length of each word in a list of strings.

```python
lst = ["hello", "World", "this is programming", "WIth", "Python"]
len_str = list(map(len, lst))
print(len_str)
```

### Q9
**Problem:** Use `filter()` along with a predefined `is_prime` helper function to extract prime numbers from a list.

```python
def is_prime(n):
    if n < 2:
        return False

    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

lst = [x for x in range(1, 101)]
prime_lst = filter(lambda x: is_prime(x), lst)
print(list(prime_lst))
```

### Q10
**Problem:** Combine `map()` and `filter()`: Given a list of numbers, filter out the odd numbers, and then use map to square the remaining even numbers.

```python
lst = [x for x in range(1, 11)]
odd_square = map(lambda x: x * x, filter(lambda x: x % 2 != 0, lst))
print(list(odd_square))
```
