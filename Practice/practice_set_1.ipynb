## 🔹 FUNCTIONS (Q1–Q8)

Q1. Write a function to check if a number is prime.

```python
# Your code here
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

print(is_prime(7))
print(is_prime(13))
print(is_prime(1))
print(is_prime(12))
print(is_prime(28))
```

Q2. Write a function to compute factorial using recursion.

```python
# Your code here

def factorial(n):
    fac = 1
    for i in range(2, n + 1):
        fac *= i
    return fac

print(factorial(5))
```

Q3. Write a function to reverse a number.

```python
# Your code here
def reverse(n):
    rev = 0
    while n != 0:
        rev = rev * 10 + n % 10
        n //= 10
    return rev

print(reverse(1234))
```

Q4. Write a function to check if a string is palindrome.

```python
# Your code here
def is_palindrome(text):
    text = text.lower()
    return text == text[::-1]

print(is_palindrome("Mom"))
print(is_palindrome("God"))
print(is_palindrome("madam"))
```

Q5. Write a function to return sum of digits of a number.

```python
# Your code here
def sum_of_digits(n):
    sum = 0
    while n != 0:
        sum += n % 10
        n //= 10
    return sum

print(sum_of_digits(1235))
```

Q6. Write a function to generate Fibonacci series up to n terms.

```python
# Your code here
def fibonacci(n):
    a, b = 0, 1
    for _ in range(n):
        print(a)
        a, b = b, a + b

fibonacci(10)
```

Q14. Create a class Student with name and marks, and display method.

```python
# Your code here
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def __str__(self):
        return f"Name: {self.name}, Marks: {self.marks}"

s1 = Student("Shaheb", 100)
s2 = Student("Aaditya", 40)

print(s1)
print(s2)
```

Q15. Create a class Rectangle to calculate area and perimeter.

```python
# Your code here
class Rectangle:
    def __init__(self, length, breadth):
        self.length = length
        self.breadth = breadth
    
    def area(self):
        return self.length * self.breadth

    def perimeter(self):
        return 2 * (self.length + self.breadth)
    
    def __str__(self):
        return f"Length: {self.length}, Breadth: {self.breadth}"

r1 = Rectangle(3, 4)
print(r1)
print(f"Area: {r1.area()}")
print(f"Perimeter: {r1.perimeter()}")
```

Q16. Create a class BankAccount with deposit and withdraw methods.

```python
# Your code here
class BankAccount:
    def __init__(self, amount):
        self.__amount = amount
    
    def withdraw(self, amount):
        if amount < self.__amount:
            self.__amount -= amount
            print("Transaction successful!")
        else:
            print("Invalid transaction.")

    def deposit(self, amount):
        if amount < 0:
            print("Invalid transaction.")
        else:
            self.__amount += amount
            print("Transaction successful!")
        
```

Q28. Count frequency of each word in a file.

```python
# Your code here
import re
freq = {}
with open("test.txt", "r") as file:
    words = re.findall(r"\b\w+\b", file.read().lower())
    for word in words:
        freq[word] = freq.get(word, 0) + 1
print(freq)
```
