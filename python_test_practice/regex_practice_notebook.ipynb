## Question 1
Extract all numbers from the following string.
`"My marks are 85 in math and 90 in physics"`
Expected output: `['85', '90']`

```python
import re
text = "My marks are 85 in math and 90 in physics"
nums = re.findall(r"\d+", text)

print(nums)
```

## Question 2
Extract all words from the string:
`"Python is fun123"`
Expected output: `['Python', 'is', 'fun']`

```python
text = "Python is fun123"
words = re.findall(r"[a-zA-Z]+", text)

print(words)
```

```python
## Question 3
Find all email addresses in the string:
`"Contact: abc@gmail.com or test123@yahoo.com"`
```

```python
text = "Contact: abc@gmail.com or test123@yahoo.com"
email_addr = re.findall(r"\w+@\w+\.\w+", text)

print(email_addr)
```

## Question 4
Check if the following string starts with the word `Hello` using regex.
`"Hello world"`

```python
text = "Hello world"
pattern = re.compile(r"^Hello")
print(pattern.match(text))
```

## Question 5
Replace all digits in the string with `#`.
`"Room 101 is on floor 5"`
Expected output: `"Room ### is on floor #"`

```python
text = "Room 101 is on floor 5"
out = re.sub(r"\d", "#", text)

print(out)
```

## Question 6
Count the number of vowels in the string using regex.
`"Hello World"`

```python
text = "Hello World"
vowels = re.findall(r"[aeiouAEIOU]", text)
print(len(vowels))
```

## Question 7
Extract all capitalised words from the sentence:
`"Rahul and Priya live in Delhi"`
Expected output: `['Rahul', 'Priya', 'Delhi']`

```python
text = "Rahul and Priya live in Delhi"
out = re.findall(r"[A-Z][a-z]+", text)
print(out)
```

## Question 8
Validate a phone number that must contain exactly **10 digits**.
Take input from the user and print `Valid` or `Invalid`.

```python
phNum = input("Phone Number: ")

if re.fullmatch(r"\d{10}", phNum):
    print("Valid")
else:
    print("Invalid")
```

## Question 9
Remove extra spaces from the string:
`"Python   is   great"`
Expected output: `"Python is great"`

```python
text = "Python   is   great"
out = re.sub(r"\s+", " ", text)
print(out)
```
