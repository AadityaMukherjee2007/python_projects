# Complete Regex Practice Notebook
Learn and practise all core regex functions step-by-step.

```python
import re

text = '''
User: john_doe
Email: john123@gmail.com
Backup Email: jd@test.org
IP: 192.168.1.1
Invalid IP: 999.999.999.999
Phone: 9876543210
Date: 12-05-2023
Another Date: 01/01/2024
'''

print(text)
```

## 1. re.search()
Find first match

```python
match = re.search(r"Email: (\S+)", text)
print(match.group() if match else 'No match')
```

## 2. re.findall()
Find all matches

```python
emails = re.findall(r"[\w.-]+@[\w.-]+", text)
print(emails)
```

## 3. re.finditer()
Iterate over matches

```python
for m in re.finditer(r"\d+", text):
    print(m.group(), m.start())
```

## 4. re.sub()
Replace matches

```python
masked = re.sub(r"\d", "*", text)
print(masked)
```

## 5. re.split()
Split string

```python
parts = re.split(r"\s+", text)
print(parts[:10])
```

## 6. re.compile()
Reuse regex

```python
pattern = re.compile(r"\b\d{4}\b")
print(pattern.findall(text))
```
