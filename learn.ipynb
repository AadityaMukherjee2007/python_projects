# Copy in python
 - ### Shallow
    A shallow copy creates a new list but does not create a copies of nested objects
    
 - ### Deep
    A deep copy creates a new list and new copies of all nested objects

Q. Write a program to check whether a given list contains any repeating elements

eg: 
- input in list [1, 2, 3, 4]
    output: False
- input in list [1, 2, 3, 2]
    output: True

```python
# arr = [1, 2, 3, 4]
arr = []

def hasDuplicates(arr):
    for i in range(0, len(arr)):
        for j in range(i + 1, len(arr)):
            if arr[i] == arr[j]:
                return True
    return False

def main():
    while True:
        try:
            n = int(input("Enter Element:"))
            arr.append(n)
        except ValueError:
            break
    
    print(arr)
    print(hasDuplicates(arr))

if __name__ == '__main__':
    main()
```

Q. Given a string find out the longest palindroming substring.

```python
def is_palindrome(s):
    return s == s[::-1]

def main():
    words = input("Sentence:").strip().split(" ")
    largest_word = words[0]
    l = len(largest_word)

    for word in words:
        if is_palindrome(word):
            if len(word) > l:
                largest_word = word
                l = len(word)

    print(f"Largest Palindrome Word: {largest_word}")

if __name__ == '__main__':
    main()
```

```python
def is_palindrome(s):
    return s == s[::-1]

def main():
    word = input("Word:").strip()
    largest_seq = []
    l = 0

    for i in range(2, len(words)):
        seq = word[0:i]
        if is_palindrome(seq):
            if len(seq) >= l:
                largest_seq.append(seq)
                l = len(seq)
    
    for i in range(0, len(words) - 1):
        seq = word[i:len(words)]
        if is_palindrome(seq):
            if len(seq) >= l:
                largest_seq.append(seq)
                l = len(seq)

    for i in range(1, len(words)):
        for j in range(1, len(words) - 1):
            seq = word[i:j]
            if is_palindrome(seq):
                if len(seq) >= l:
                    largest_seq.append(seq)
                    l = len(seq)

    print(f"Largest Palindrome Seq: {set(largest_seq)}")

if __name__ == '__main__':
    main()
```

You are given a list of student records where student = [("701", "std1", 50), ...]
Write a user defined function named is_eligible that returns true if the student's mark is less than 30. The list is passed as an argument.
Use the above function with filter operation to extract the ids of the eligible students.
Using lambda function with map create a new list with student's id and update it's marks where students get a grace marks of 5 if they get 30. 

```python
students = [
    ("701", "std1", 57), 
    ("702", "std2", 51), 
    ("703", "std3", 28), 
    ("704", "std4", 33), 
    ("705", "std5", 30), 
    ("706", "std6", 12)
]


def is_eligible(rec):
    return rec[2] < 30

eligible = list(filter(is_eligible, students))
print(eligible)

updated_marks = list(map(lambda x:x[2]+5, eligible))
print(updated_marks)
```
